## Grouping:

```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv("data.csv")

df['salary'].unique()

df['salary']=pd.to_numeric(df['salary'],errors='coerce')

#mean of ages of person in different salary
# df.groupby('city')['age'].mean()
# df.groupby('city')['salary'].mean()

df.groupby("department").agg({
    "salary": ["mean", "max"],
    "employee_id": "count"
})
```
